```python

import folium
import imgkit
import matplotlib.pyplot as plt
import matplotlib.image as mpimg


def create_austin_image():
    # Create a Folium map centered on Austin
    m = folium.Map(location=[30.2672, -97.7431], zoom_start=12,width='100%', height='100%')

    # Save the map to an HTML file
    m.save('austin_map.html')
    imgkit.from_file('austin_map.html', 'austin_map_example.png')

def test_plotting():
    # Load the image
    img = mpimg.imread('austin.png')

    # Create a figure and axis
    fig, ax = plt.subplots(figsize=(10, 10))

    # Display the image
    ax.imshow(img, extent=[-97.8, -97.6, 30.2, 30.3])  # Adjust the extent based on your map bounds

    # Now plot your points on top of the map
    points = [(30.25, -97.75), (30.24, -97.76), (30.23, -97.77)]
    labels = ['A', 'B', 'C']

    for (lat, lon), label in zip(points, labels):
        ax.scatter(lon, lat, c='blue', alpha=0.7)
        ax.annotate(label, (lon, lat), textcoords="offset points", xytext=(0, 10), ha='center')

    plt.show()

create_austin_image()
test_plotting()
```
